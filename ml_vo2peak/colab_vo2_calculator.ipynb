# VO₂peak Response Calculator
### Post-COVID rehabilitation (8-week programs)

Estimates a patient's chance of a **clinical improvement in peak oxygen
uptake (Δ ≥ 1.2 mL·kg⁻¹·min⁻¹)** under three scenarios.

**Model & results** — elastic-net logistic regression (compact-5); person-grouped,
stratified, repeated 5-fold cross-validation: **F1 0.86 ± 0.09, AUC 0.92 ± 0.07**;
nested CV AUC 0.85 (95% CI 0.76–0.99).

1. Press ▶ next to **Step 1** and wait for the ✓.
2. Fill in **every** Step 2 field (possible range next to each field).
3. Press ▶ next to **Step 2** to see the result.

## What the inputs mean

| Field | Possible | Studied range | Notes |
|---|---|---|---|
| Steps per day | 0–150 000 | 791–12 264 | activity-tracker daily step count, **before** the program |
| Physical-activity duration | 0–1440 min | 0–209 min | daily active time from the tracker |
| Walk distance (6-MWT) | 0–1500 m | 279–640 m | six-minute walk test, **before** the program |
| Regular exercise habit | Yes / No | Yes / No | structured exercise habit at baseline |

All values are **baseline** (pre-program) measurements. Every physiologically
possible entry is computed; a value outside the possible range (e.g. an
activity duration above 1440 min/day) is a data-entry error, so the probability
of success is shown as "?". Inside the possible range but outside the
studied range, the estimate is an extrapolation: a caution banner is shown and
the bar chart and frequency wording are omitted. Every estimate carries a 95%
confidence interval from a person-cluster bootstrap (B = 500).

## Limitations
Exploratory model: the feature set was found by a recorded, multiplicity-charged
search on the same 53-record cohort and has **not** been prospectively
validated. Confidence intervals reflect sampling uncertainty of this small
cohort only. Estimates reflect one single-center study and do not replace
clinical judgement.
