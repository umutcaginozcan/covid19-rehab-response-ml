# 6-Minute Walk Test — Response Calculator
### Post-COVID rehabilitation (8-week programs)

Estimates a patient's chance of a **clinically meaningful improvement**
(≥ 25 m in six-minute walk distance) under three scenarios.

**Model & results** — logistic regression (compact-8); person-grouped,
stratified, repeated 5-fold cross-validation: **F1 0.80 ± 0.09, AUC 0.87 ± 0.08**.

**How to use — no coding needed:**
1. Press ▶ next to **Step 1** and wait for the ✓.
2. Fill in **every** Step 2 field (possible range next to each field).
3. Press ▶ next to **Step 2** to see the result.

## What the inputs mean

| Field | Possible | Studied range | Notes |
|---|---|---|---|
| Age | 0–120 y | 38–74 y | |
| BMI | 5–200 | 19–38 kg/m² | body-mass index |
| Walk distance (6-MWT) | 0–1500 m | 250–640 m | six-minute walk test, **before** the program |
| MIP | 0–400 | 34–145 cmH₂O | maximal inspiratory pressure |
| VO₂peak | 1–100 | 7–28 mL·kg⁻¹·min⁻¹ | from cardiopulmonary exercise test |
| FSS | 9–63 | 18–63 | Fatigue Severity Scale total (9 items × 1–7) |
| SGRQ | 0–100 | 7–91 | St. George's Respiratory Questionnaire total |

All values are **baseline** (pre-program) measurements. Every physiologically
possible entry is computed; a value outside the possible range (e.g. FSS 65 —
the scale maxes at 63) is a data-entry error, so the probability of success is
shown as "?". Inside the possible range but outside the
studied range, the estimate is an extrapolation: a caution banner is shown and
the bar chart and frequency wording are omitted. Every estimate carries a 95%
confidence interval from a person-cluster bootstrap (B = 1000).
*Explain result for* selects the scenario explained in the "Why" panel.

## Limitations
Estimates come from a logistic-regression model trained on 55 records of
49 adults with post-COVID-19 syndrome in one center; internal cross-validation
AUC ≈ 0.87. Confidence intervals reflect sampling uncertainty of this small
cohort only. Results may not transfer to other populations and do not replace
clinical judgement.
