### Post-COVID rehabilitation (8-week programs)

(≥ 25 m in six-minute walk distance) under three scenarios.

**How to use — no coding needed:**
1. Press ▶ next to **Step 1** and wait for the ✓.
2. Type the patient's values into the **Step 2** form.
3. Press ▶ next to **Step 2** to see the result.

```python
#@title Step 2 — Enter the patient\'s baseline values, then press ▶ { display-mode: "form" }
Age_years = 59  #@param {type:"number"}
BMI_kg_m2 = 28.0  #@param {type:"number"}
Walk_distance_6MWT_meters = 499  #@param {type:"number"}
MIP_cmH2O = 84  #@param {type:"number"}
VO2peak_mL_kg_min = 19.0  #@param {type:"number"}
FSS_fatigue_score = 50  #@param {type:"number"}
SGRQ_total_score = 45  #@param {type:"number"}
Explain_result_for = "IMT"  #@param ["IMT", "HIIT", "No structured rehabilitation"]

from IPython.display import HTML, display
from math import factorial

INK, MUTED, FAINT, LINE = '#1f2937', '#6b7280', '#9ca3af', '#eef0f3'
FONT = "-apple-system,'Segoe UI',Roboto,Arial,sans-serif"

# studied population (cohort min–max); outside -> extrapolation warning
_checks = [
    ('Age (years)', Age_years, 38, 74),
    ('BMI (kg/m²)', BMI_kg_m2, 19, 38),
    ('6-MWT walk distance (m)', Walk_distance_6MWT_meters, 250, 640),
    ('MIP (cmH₂O)', MIP_cmH2O, 34, 145),
    ('VO₂peak (mL/kg/min)', VO2peak_mL_kg_min, 7, 28),
    ('FSS fatigue score', FSS_fatigue_score, 18, 63),
    ('SGRQ total score', SGRQ_total_score, 7, 91),
]
_problems = [f'{n}: studied range {lo}–{hi}, entered {v:g}'
             for n, v, lo, hi in _checks if not (lo <= v <= hi)]
warn = ''
if _problems:
    items = ''.join(f'<li style="margin:3px 0;">{p}</li>' for p in _problems)
    warn = f"""<div style="max-width:660px;background:#fffbeb;border:1px solid #fde68a;
      border-left:4px solid #f59e0b;border-radius:10px;padding:12px 18px;margin-bottom:12px;
      font-family:{FONT};color:#78350f;font-size:13px;">
      <b>Outside the studied population — extrapolated estimate, interpret with caution.</b>
      <ul style="margin:6px 0 0 18px;padding:0;">{items}</ul></div>"""

probs = predict_all(Age_years, BMI_kg_m2, Walk_distance_6MWT_meters,
                    MIP_cmH2O, VO2peak_mL_kg_min, FSS_fatigue_score, SGRQ_total_score)

# exact Shapley vs cohort-median reference, under the chosen scenario
_arm_key = {'IMT': 'Inspiratory muscle training (IMT)',
            'HIIT': 'High-intensity interval training (HIIT)',
            'No structured rehabilitation': 'No structured rehabilitation'}[Explain_result_for]
gi, gh = ARMS[_arm_key]
REF = [499, 84, 19, 50, 45, 59, 28]
CLIN_IDX = [2, 3, 4, 5, 6, 7, 8]
FEATS = [('6-MWT distance', f'{Walk_distance_6MWT_meters:g} m'), ('MIP', f'{MIP_cmH2O:g} cmH₂O'),
         ('VO₂peak', f'{VO2peak_mL_kg_min:g}'), ('FSS', f'{FSS_fatigue_score:g}'),
         ('SGRQ', f'{SGRQ_total_score:g}'), ('Age', f'{Age_years:g}'), ('BMI', f'{BMI_kg_m2:g}')]
entered = [Walk_distance_6MWT_meters, MIP_cmH2O, VO2peak_mL_kg_min,
           FSS_fatigue_score, SGRQ_total_score, Age_years, BMI_kg_m2]
n = 7
masks = np.array([[(m >> i) & 1 for i in range(n)] for m in range(2 ** n)])
Xm = np.tile([gi, gh] + REF, (2 ** n, 1)).astype(float)
for i in range(n):
    Xm[masks[:, i] == 1, CLIN_IDX[i]] = entered[i]
fv = MODEL.predict_proba(Xm)[:, 1]
phi = np.zeros(n)
for mask in range(2 ** n):
    s = int(masks[mask].sum())
    for i in range(n):
        if not masks[mask][i]:
            w = factorial(s) * factorial(n - s - 1) / factorial(n)
            phi[i] += w * (fv[mask | (1 << i)] - fv[mask])
base_p, final_p = fv[0] * 100, fv[-1] * 100

rows = ''
for (arm, p), color in zip(probs.items(), ARM_COLORS):
    n10 = 'fewer than 1' if p < 0.1 else 'about ' + str(round(p * 10))
    rows += f"""<div style="display:flex;align-items:center;gap:14px;margin:11px 0;">
      <div style="width:252px;font-size:13.5px;color:{INK};line-height:1.25;">{arm}
        <div style="font-size:11px;color:{MUTED};">{n10} of 10 similar patients</div></div>
      <div style="flex:1;height:12px;background:{LINE};border-radius:6px;">
        <div style="width:{min(p*100,100):.0f}%;height:100%;background:{color};border-radius:6px;"></div></div>
      <div style="width:52px;text-align:right;font-size:16px;font-weight:700;color:{INK};">{p*100:.0f}%</div>
    </div>"""

pp = phi * 100
POS, NEG = '#FF0D57', '#1E88E5'
order = np.argsort(-np.abs(pp))
big = [i for i in order if abs(pp[i]) >= 0.5]
wf = ''
if big:
    cum = base_p
    for i in big:
        v = pp[i]
        w = abs(v)
        left = cum if v >= 0 else cum + v
        cum += v
        color = POS if v >= 0 else NEG
        chev = ('polygon(0 0, calc(100% - 7px) 0, 100% 50%, calc(100% - 7px) 100%, 0 100%)'
                if v >= 0 else 'polygon(7px 0, 100% 0, 100% 100%, 7px 100%, 0 50%)')
        inside = w >= 14
        txt = f'{v:+.0f}%'
        bar_txt = (f'<span style="color:#ffffff;font-size:11.5px;font-weight:700;">{txt}</span>' if inside else '')
        out_txt = ('' if inside else
                   f'<div style="position:absolute;top:1px;font-size:11.5px;font-weight:700;color:{color};'
                   + (f'left:calc({min(left + w, 97):.1f}% + 5px);">' if v >= 0 else f'right:calc({100 - max(left, 3):.1f}% + 5px);">')
                   + txt + '</div>')
        wf += f"""<div style="display:flex;align-items:center;gap:12px;margin:5px 0;">
          <div style="width:190px;font-size:12.5px;text-align:right;color:{INK};">
            <span style="color:{FAINT};">{FEATS[i][1]} = </span>{FEATS[i][0]}</div>
          <div style="position:relative;flex:1;height:17px;">
            <div style="position:absolute;left:{min(max(base_p,0),100):.1f}%;top:-5px;bottom:-5px;width:1px;border-left:1px dashed #cbd5e1;"></div>
            <div style="position:absolute;left:{min(max(left,0),99):.1f}%;width:{max(min(w,99),1.2):.1f}%;height:17px;background:{color};
                 clip-path:{chev};display:flex;align-items:center;justify-content:center;">{bar_txt}</div>
            {out_txt}
          </div></div>"""
    endx = min(max(cum, 8), 92)
    wf = (f'<div style="display:flex;gap:12px;margin-bottom:2px;"><div style="width:190px;"></div>'
          f'<div style="position:relative;flex:1;height:15px;font-size:11.5px;color:{INK};">'
          f'<div style="position:absolute;left:{endx:.1f}%;transform:translateX(-50%);font-weight:700;white-space:nowrap;">this patient: {final_p:.0f}%</div></div></div>') + wf
    basex = min(max(base_p, 8), 92)
    wf += (f'<div style="display:flex;gap:12px;margin-top:2px;"><div style="width:190px;"></div>'
           f'<div style="position:relative;flex:1;height:15px;font-size:11px;color:{MUTED};border-top:1px solid #e5e7eb;">'
           f'<div style="position:absolute;left:{basex:.1f}%;transform:translateX(-50%);white-space:nowrap;">typical patient: {base_p:.0f}%</div></div></div>')
    small = [i for i in order if abs(pp[i]) < 0.5]
    if small:
        wf += f'<div style="font-size:11.5px;color:{FAINT};margin-top:6px;">Other values: ≈ no effect</div>'
else:
    wf = f'<div style="font-size:13px;color:{MUTED};margin:10px 0 4px;">All entered values are close to the typical study patient — no individual differences to show.</div>'

display(HTML(f"""{warn}<div style="max-width:660px;background:#ffffff;border:1px solid #e5e7eb;border-radius:14px;
  padding:22px 26px 18px;font-family:{FONT};color:{INK};box-shadow:0 1px 3px rgba(0,0,0,.08);">
  <div style="font-size:15.5px;font-weight:700;">Chance of meaningful improvement</div>
  <div style="font-size:12px;color:{MUTED};margin:2px 0 10px;">≥ 25 m gain in six-minute walk distance after the 8-week program</div>
  {rows}
  <hr style="border:none;border-top:1px solid {LINE};margin:18px 0 14px;">
  <div style="font-size:13.5px;font-weight:700;">Why — under {Explain_result_for}</div>
  <div style="font-size:12px;color:{MUTED};margin:2px 0 10px;">
    How each entered value moves this patient from the typical study patient.
    Red pushes the chance up, blue pushes it down.</div>
  {wf}
  <hr style="border:none;border-top:1px solid {LINE};margin:16px 0 10px;">
  <div style="font-size:11px;color:{FAINT};line-height:1.5;">Research prototype — not medical advice.
    Logistic-regression model from a single-center study (55 records); internal cross-validation AUC ≈ 0.87.
    The embedded model contains aggregate coefficients only — no patient-level data.</div>
</div>"""))
```

## What the inputs mean

| Field | Studied range | Notes |
|---|---|---|
| Age | 38–74 years | |
| BMI | 19–38 kg/m² | body-mass index |
| Walk distance (6-MWT) | 250–640 m | six-minute walk test, **before** the program |
| MIP | 34–145 cmH₂O | maximal inspiratory pressure |
| VO₂peak | 7–28 mL·kg⁻¹·min⁻¹ | from cardiopulmonary exercise test |
| FSS | 18–63 | Fatigue Severity Scale total |
| SGRQ | 7–91 | St. George's Respiratory Questionnaire total |

All values are **baseline** (pre-program) measurements. Outside the studied
ranges the tool still estimates (the model extrapolates smoothly) but shows a
caution banner. *Explain result for* selects the scenario explained in the
"Why" panel.

## Limitations
Estimates come from a logistic-regression model trained on 55 records of
49 adults with post-COVID-19 syndrome in one center; internal cross-validation
AUC ≈ 0.87. Results may not transfer to other populations and do not replace
clinical judgement.
